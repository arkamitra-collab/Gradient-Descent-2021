```python
import math

# AND gate dataset
X = [
    [0,0],
    [0,1],
    [1,0],
    [1,1]
]

Y = [0,0,0,1]

# Sigmoid activation
def sigmoid(z):
    return 1/(1 + math.exp(-z))
```

```python
# Initialize weights and bias
w1 = 0.2
w2 = -0.3
b  = 0.4

learning_rate = 0.5
epochs = 30000
```

```python
# Training using gradient descent
for epoch in range(epochs):
    total_loss = 0

    for i in range(len(X)):
        x1,x2 = X[i]
        y = Y[i]

        # Forward pass
        z = w1*x1 + w2*x2 + b
        y_hat = sigmoid(z)

        # Loss
        error = y_hat - y
        total_loss += 0.5*(error**2)

        
        
        # Gradients
        delta = error * y_hat*(1-y_hat)

        dw1 = delta*x1
        dw2 = delta*x2
        db  = delta

        # Gradient Descent update
        w1 = w1 - learning_rate*dw1
        w2 = w2 - learning_rate*dw2
        b  = b  - learning_rate*db

        if epoch % 1000 == 0:
            print("Epoch:",epoch,"Loss:",total_loss)

        

    
```

```python
# Testing
print("\nFinal Weights:")
print(w1,w2,b)

```

```python
print("\nAND Gate Output:")
for x1,x2 in X:
    z = w1*x1 + w2*x2 + b
    y_hat = sigmoid(z)

    # Threshold at 0.5
    output = 1 if y_hat>=0.5 else 0

    print([x1,x2],"->",output,"(",round(y_hat,4),")")
```
